# Exploratory Data Analysis (EDA)

### Total Deliveries by Year

```python
yearly_deliveries = df.groupby('Year')['Estimated_Deliveries'].sum()
yearly_deliveries
```

# Logical Filtering (AND, OR, NOT)

### AND Condition

Deliveries after 2020 AND Model Y only


```python
df_and = df[(df['Year'] > 2020) & (df['Model'] == 'Model Y')]
df_and
```

### OR Condition

Model 3 OR Model Y deliveries


```python
df_or = df[(df['Model'] == 'Model 3') | (df['Model'] == 'Model Y')]
df_or
```

### NOT Condition

Exclude Model S

```python
df_not = df[~(df['Model'] == 'Model S')]
df_not
```

### Combined Conditions

Model Y in USA after 2021

```python
df_combo = df[(df['Model'] == 'Model Y') & (df['Region'] == 'USA') & (df['Year'] > 2021)]
```

# Data Visualization (Graphs)

### Line Graph – Yearly Deliveries Trend

```python
yearly_deliveries.plot(kind='line', marker='o')
plt.title('Tesla Yearly Deliveries (2015–2025)')
plt.xlabel('Year')
plt.ylabel('Total Deliveries')
plt.grid()
plt.show()
```

### Bar Chart – Model Wise Deliveries

```python
model_deliveries = df.groupby('Model')['Estimated_Deliveries'].sum()
model_deliveries.plot(kind='bar')
plt.title('Total Deliveries by Tesla Model')
plt.xlabel('Model')
plt.ylabel('Deliveries')
plt.show()
```

### Pie Chart – Market Share by Model

```python
plt.pie(model_deliveries, labels=model_deliveries.index, autopct='%1.1f%%')
plt.title('Tesla Model Market Share')
plt.show()
```

###  Box Plot – Delivery Distribution

```python
sns.boxplot(x='Model', y='Estimated_Deliveries', data=df)
plt.title('Delivery Distribution by Model')
plt.show()
```

### Heatmap – Deliveries by Year & Model

```python
pivot = df.pivot_table(values='Estimated_Deliveries', index='Year', columns='Model', aggfunc='sum')
sns.heatmap(pivot, annot=True, fmt='g', cmap='coolwarm')
plt.title('Heatmap of Tesla Deliveries')
plt.show
```
